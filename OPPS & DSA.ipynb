```python
#FOR UPDATING THE ARRAY WE WILL DO IT BY RANGE FUNCTION

def inc(a):
    for i in range(len(a)):
         a[i]=a[i]+10
    # print(a)
    return a
   
```

```python
n=int(input("Enter the size of list:"))
k=[]
for i in range(n):
    v=int(input("Enter values:"))
    k.append(v)
print(k)
print(inc(k)) 
```

```python
# W.A.P TO RETURN THE LARGEST NUMBER FROM A TUPLE USING FUNCTION

def typ(t):
    li=list(t)
    for i in range(len(t)):
        if(li[i]>li[i+1]):
            temp=li[i]
            li[i]=li[i+1]
            li[i+1]=temp
        t=tuple(li)
        print(t)
        print(temp)
```

```python
n=int(input("Enter size:"))
t=()
for i in range(n):
    v=int(input("Enter value"))
    t=v
print(t)
typ(t)
```

```python
# W.A.P TO RETURN VOWEL FROM STRING USING FUNCTION

def vwl(a):
    k=[]
    for i in a:
        if (i=="a" or i=="A" or i=="e" or i=="E" or i=="i" or i=="I" or i=="o" or i=="O" or i=="u" or i=="U"):
            k.append(i)
    return k
        
```

```python
n=input("Enter String:")
print(vwl(n))
```

```python
# W.A.P TO RETURN THE LARGEST NUMBER FROM A TUPLE USING FUNCTION

def mx(t):
    l=t[0]
    for i in t:
        if i>l:
            l=i
    return l
```

```python
k=(34,56,76,43)
g=mx(k)
print(g)
```

```python
# W.A.P TO RETURN THE SMALLEST NUMBER FROM A TUPLE USING FUNCTION

def sml(t):
    l=t[0]
    for i in t:
        if i<l:
            l=i
    return l
```

```python
k=(34,56,76,43)
g=sml(k)
print(g)
```

```python
# W.A.P TO RETURN VOWEL FROM STRING USING FUNCTION

def vwl(a):
    k=[]
    for i in a:
        if (i=="a" or i=="A" or i=="e" or i=="E" or i=="i" or i=="I" or i=="o" or i=="O" or i=="u" or i=="U"):
            k.append(i)
    return k
        
```

```python
n=input("Enter String:")
print(vwl(n))
```

```python
# WAP TO PRINT LARGEST AND SMALLEST IN AN ARRAY

def comp(t):
    min=t[0]
    max=t[0]
    for i in range(len(t)):
        if t[i]>max:
            max=t[i]
        elif t[i]<min:
            min=t[i]
    print(min)
    print(max)
```

```python
k=(2,34,56,7)
comp(k)
print(k)
```

```python
# A to Z =(65-90)
# a to z =(97-122)

def utol(c):
    k=ord(c)
    print(k)
    print(chr(k+32))

def ltou(c):
    k=ord(c)
    print(chr(k-32))
```

```python
utol("S")
ltou("a")
```

## CLASS

```python
# TO SECURE FUNCTION WE MAKE CLASS

# CLASS 
# IT IS A BLUE PRINT OF OBJECT
# IT IS A COLLECTION OF DATA MEMBER AND MEMBER FUNCTION
# IT IS CALLED USER DEFINED DATA TYPE

#DATA MEMBER IS A VARIABLE
# WE CAN ACCES A DATA MEMBER OF A CLASS USING OBJECT

# LIKE
# a=10
# b=20

# BY USING SELF WE CAN DETECT THE OBJECT OF THE CURRENT CLASS 

# WHENEVER WE DECLARE A VARIABLE INSIDE A FUNCTION IT IS KNOWN AS INSTANCE VARIABLE EXCLUDING OF SELF LIKE def mysum(self,x,y):
# WHENEVER WE DECLARE A VARIABLE INSIDE A CLASS IT IS KNOWN AS CLASS VARIABLE
# WHENEVER WE CREATE A FUNCTION WE WRITE SELF INSIDE THE FUNCTION AS AN OBJECT

# WITHOUT SELF WE CAN CALL LIKE BUT WE DONT NEED TO CREATE AN OBJECT WE WILL CALL BY USING THE CLASS NAME
# class cybrom:
#     def show():
#         print("Hii")
# cybrom.show()
```

```python
# SMALLEST PROGRAM OF CLASS
# class cybrom:
#     pass


class cybrom:
    a=10
    b=30    
    def sum(self):
        print(self.a+self.b)
    def sub(self,x,y):
        print(x-y)
    def mul(self):
        print(self.a*self.b)
```

```python
d=cybrom()
print(d.a+d.b)
d.sum()
d.sub(12,34)
d.mul()
```

```python
# PYTHON DOESNT SUPPORT FUNCTION OVERLOADING

#INHERITENCE

#SINGLE INHERITENCE WHEN THERE IS ONLY ONE PARENT AND CHILD

class parents:
    def st1(self):
        print("Parents")
class childs(parents):
    def st2(self):
        print("Child")        
```

```python
c=childs()
c.st1()
c.st2()
```

```python
#MULTILEVEL INHERITENCE

class grandfather:
    def g1(self):
        print("GrandFather")
class father(grandfather):
    def f1(self):
        print("Father")   
class child(father):
    def c1(self):
        print("child")                
```

```python
c=child()
c.c1()
c.f1()
c.g1()
```

```python
#MULTIPLE

class tenth:
    def t1(self):
        print("10th")
class twelth:
    def t2(self):
        print("12th")   
class graduation(tenth,twelth):
    def g1(self):
        print("Graduated!!")     
```

```python
g=graduation()
g.t1()
g.t2()
g.g1()
```

```python
class student:
    def __init__(self):
        self.cybrom=[]
    def ad(self,data):
        self.cybrom.append(data)
    def show(self):
        print(self.cybrom)
```

```python
s=student()
s.ad(200)
s.ad(300)
s.show()
```

```python
# PARAMETERIZED CONSTRUCTOR

class college:
    x,y=0,0
    def __init__(self,a,b):
        print("a:",a)
        print("b:",b)
        self.x=a
        self.y=b
    def sums(self):
        print(self.x+self.y)
        
```

```python
obj=college(34,56)
obj.sums()
```
